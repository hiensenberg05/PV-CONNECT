```python
import requests
import base64
import os
from dotenv import load_dotenv

# 1. Load API Key
load_dotenv()
api_key = os.getenv("GOOGLE_CLOUD_VISION") 

# 2. Define the Image URL (Scones)
image_url = "https://storage.googleapis.com/generativeai-downloads/images/scones.jpg"

# 3. Construct the Request for LABEL_DETECTION
# We need to send the image content or URL. Sending URL here for simplicity.
url = f"https://vision.googleapis.com/v1/images:annotate?key={api_key}"

payload = {
  "requests": [
    {
      "image": {
        "source": {
          "imageUri": image_url
        }
      },
      "features": [
        {
          "type": "LABEL_DETECTION",
          "maxResults": 5
        }
      ]
    }
  ]
}

# 4. Make the Request
try:
    print("Sending request to Google Cloud Vision API...")
    response = requests.post(url, json=payload)
    
    if response.status_code == 200:
        results = response.json()
        print("\n✅ Success! Labels found:")
        labels = results.get('responses', [])[0].get('labelAnnotations', [])
        for label in labels:
            print(f"- {label['description']} ({label['score']:.2%})")
    else:
        print(f"\n❌ Error {response.status_code}: {response.text}")

except Exception as e:
    print(f"\n❌ Exception: {e}")
```

```python
import httpx
import base64
import requests
from dotenv import load_dotenv
import os

load_dotenv()

async def test_cloud_vision():
    # 1. Get Key
    api_key = os.getenv("GOOGLE_CLOUD_VISION")
    if not api_key:
        print("❌ Error: GOOGLE_CLOUD_VISION key not found in .env")
        return

    # 2. Get Test Image (Scones)
    img_url = "https://storage.googleapis.com/generativeai-downloads/images/scones.jpg"
    img_data = requests.get(img_url).content
    b64_image = base64.b64encode(img_data).decode("utf-8")

    # 3. Call Google Cloud Vision API
    url = f"https://vision.googleapis.com/v1/images:annotate?key={api_key}"
    payload = {
        "requests": [
            {
                "image": {"content": b64_image},
                "features": [
                    {"type": "LABEL_DETECTION", "maxResults": 5},
                    {"type": "TEXT_DETECTION"}
                ]
            }
        ]
    }
    
    print("Sending request to Google Cloud Vision API...")
    async with httpx.AsyncClient() as client:
        response = await client.post(url, json=payload)
        
    if response.status_code == 200:
        print("\n✅ Success!")
        print(response.json())
    else:
        print(f"\n❌ Error {response.status_code}: {response.text}")

# Run the async test
await test_cloud_vision()
```
